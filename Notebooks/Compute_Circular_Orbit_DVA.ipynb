```python
%matplotlib inline
import matplotlib.pyplot as plt
%cd ..
```

```python
import numpy as np
import healpy as hp
from DVA_compute import aberration_sh_rotator, planar_orbit_velocity_direction
```

```python
#These examples will use Healpy coordinates, but the function should work for any arrays of thetas and phis.
nside = 64
thetas, phis = hp.pix2ang(nside, np.arange(12*nside**2))
```

```python
#Example 1. Compute DVA for velocity in the north pole direction.

DVA_computer = aberration_sh_rotator(beta=10**-4, order=3)
#To compute the DVA, use method rotate_to_velocity direction.
#First 2 arguments are the velocity theta and phi directions in radians (scalars). For north pole direction, both are zero.
#The second 2 arguments are the angles for which to compute the DVA.
DVA_computer.rotate_to_velocity_direction(0, 0, thetas, phis)

#Plot the ell=1 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell1_theta_hat, title = r'$\hat{\theta}$ 1st order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell1_phi_hat, title = r'$\hat{\phi}$ 1st order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell1_magnitude, title = r'Total 1st order aberration magnitude',
           unit = 'radians')

#Plot the ell=2 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell2_theta_hat, title = r'$\hat{\theta}$ 2nd order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell2_phi_hat, title = r'$\hat{\phi}$ 2nd order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell2_magnitude, title = r'Total 2nd order aberration magnitude',
           unit = 'radians')

#Plot the ell=3 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell3_theta_hat, title = r'$\hat{\theta}$ 3rd order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell3_phi_hat, title = r'$\hat{\phi}$ 3rd order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell3_magnitude, title = r'Total 3rd order aberration magnitude',
           unit = 'radians')

#Plot the total aberration up to order 3. 
#Plot will look basically like the order 1 aberration,
#but the high order terms may be bigger than gravitational waves.
hp.mollview(DVA_computer.full_rotated_aberration_theta_hat, title = r'$\hat{\theta}$ full aberration up to 3rd order', 
            unit = 'radians')
hp.mollview(DVA_computer.full_rotated_aberration_phi_hat, title = r'$\hat{\phi}$ full aberration up to 3rd order',
           unit = 'radians')
hp.mollview(DVA_computer.full_rotated_aberration_magnitude, title = r'Full aberration magnitude up to 3rd order',
           unit = 'radians')
```

```python
#Example 2. Compute DVA for velocity in the theta=pi/2, phi=0 direction.

#To compute the DVA, use method rotate_to_velocity direction.
#First 2 arguments are the velocity theta and phi directions in radians (scalars). 
#The second 2 arguments are the angles for which to compute the DVA.
DVA_computer.rotate_to_velocity_direction(np.pi/2, 0, thetas, phis)

#Plot the ell=1 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell1_theta_hat, title = r'$\hat{\theta}$ 1st order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell1_phi_hat, title = r'$\hat{\phi}$ 1st order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell1_magnitude, title = r'Total 1st order aberration magnitude',
           unit = 'radians')

#Plot the ell=2 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell2_theta_hat, title = r'$\hat{\theta}$ 2nd order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell2_phi_hat, title = r'$\hat{\phi}$ 2nd order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell2_magnitude, title = r'Total 2nd order aberration magnitude',
           unit = 'radians')

#Plot the ell=3 aberration.
hp.mollview(DVA_computer.rotated_aberration_ell3_theta_hat, title = r'$\hat{\theta}$ 3rd order aberration', 
            unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell3_phi_hat, title = r'$\hat{\phi}$ 3rd order aberration',
           unit = 'radians')
hp.mollview(DVA_computer.rotated_aberration_ell3_magnitude, title = r'Total 3rd order aberration magnitude',
           unit = 'radians')

#Plot the total aberration up to order 3. 
#Plot will look basically like the order 1 aberration,
#but the high order terms may be bigger than gravitational waves.
hp.mollview(DVA_computer.full_rotated_aberration_theta_hat, title = r'$\hat{\theta}$ full aberration up to 3rd order', 
            unit = 'radians')
hp.mollview(DVA_computer.full_rotated_aberration_phi_hat, title = r'$\hat{\phi}$ full aberration up to 3rd order',
           unit = 'radians')
hp.mollview(DVA_computer.full_rotated_aberration_magnitude, title = r'Full aberration magnitude up to 3rd order',
           unit = 'radians')
```

```python
#Example 3. Compute angles for a circular Earth-ish orbit in the Ecliptic plane.

test_circular_ecliptic_orbit = planar_orbit_velocity_direction()
#Set the period to 1 year.
orbital_period_seconds = 3.154e+7
#Condider observing once every 100 seconds during a year.
observation_times = np.arange(0,orbital_period_seconds, 100)
#Set orbital phase to 0. This just determines where in the Ecliptic plane the orbit stars.
orbital_phase = 0
test_circular_ecliptic_orbit.compute_circular_orbit_theta_phi(observation_times, 
                                                              orbital_period_seconds, orbital_phase)


#Visualize the orbit.
#First, show orbital path in healpy coordinates.
test_orbital_velocity_pixels = hp.vec2pix(64, test_circular_ecliptic_orbit.circular_orbit_vectors[0,:],
                                         test_circular_ecliptic_orbit.circular_orbit_vectors[1,:],
                                         test_circular_ecliptic_orbit.circular_orbit_vectors[2,:])


test_hp_orbit_pixels = np.zeros(12*64**2)
test_hp_orbit_pixels[test_orbital_velocity_pixels] = 1

hp.mollview(test_hp_orbit_pixels)
plt.show()

#Next, show theta and phi of orbit over time.
plt.plot(observation_times, 180/np.pi*test_circular_ecliptic_orbit.circular_orbit_phis, label = 'Theta')
plt.plot(observation_times, 180/np.pi*test_circular_ecliptic_orbit.circular_orbit_thetas, label = 'Phi')
plt.xlabel('Seconds')
plt.ylabel('Angle (degrees)')
plt.title('Earth orbital velocity direction in Galactic coordinates')
plt.show()
```

```python
#Example 3, continued.
#Now, compute how the Vector spherical harmonic coefficients will change over the orbit.

DVA_computer.generate_scalar_spherical_harmonics(test_circular_ecliptic_orbit.circular_orbit_thetas,
                                                 test_circular_ecliptic_orbit.circular_orbit_phis)

#Plot the ell=1 components over time during the orbit.
plt.plot(observation_times, DVA_computer.y10_initial*DVA_computer.Y10, label = r'$\Psi_{10}$')
plt.plot(observation_times, DVA_computer.y10_initial*DVA_computer.Y11, label = r'$\Psi_{11}$')
plt.plot(observation_times, DVA_computer.y10_initial*DVA_computer.Y1n1, label = r'$\Psi_{1-1}$')
plt.legend()
plt.title(r'$\ell=1$ vector spherical harmonic coefficieints')
plt.xlabel('Seconds')
plt.ylabel('Amplitude (radians)')
plt.show()

#Plot the ell=2 components over time during the orbit.
plt.plot(observation_times, DVA_computer.y20_initial*DVA_computer.Y20, label = r'$\Psi_{20}$')
plt.plot(observation_times, DVA_computer.y20_initial*DVA_computer.Y21, label = r'$\Psi_{21}$')
plt.plot(observation_times, DVA_computer.y20_initial*DVA_computer.Y2n1, label = r'$\Psi_{2-1}$')
plt.plot(observation_times, DVA_computer.y20_initial*DVA_computer.Y22, label = r'$\Psi_{22}$')
plt.plot(observation_times, DVA_computer.y20_initial*DVA_computer.Y2n2, label = r'$\Psi_{2-2}$')
plt.title(r'$\ell=2$ vector spherical harmonic coefficieints')
plt.legend()
plt.xlabel('Seconds')
plt.ylabel('Amplitude (radians)')
plt.show()

#Compute fourier transform of spherical harmonic coefficients during the orbit.
psi10_fft_time = 2*np.fft.rfft(DVA_computer.y10_initial*DVA_computer.Y10)/(observation_times.size)
psi1n1_fft_time = 2*np.fft.rfft(DVA_computer.y10_initial*DVA_computer.Y1n1)/(observation_times.size)
psi11_fft_time = 2*np.fft.rfft(DVA_computer.y10_initial*DVA_computer.Y11)/(observation_times.size)
psi21_fft_time = 2*np.fft.rfft(DVA_computer.y20_initial*DVA_computer.Y21)/(observation_times.size)
psi2n1_fft_time = 2*np.fft.rfft(DVA_computer.y20_initial*DVA_computer.Y2n1)/(observation_times.size)
psi20_fft_time = 2*np.fft.rfft(DVA_computer.y20_initial*DVA_computer.Y20)/(observation_times.size)
freqs = np.fft.rfftfreq(observation_times.size, d=(3.154e7)/observation_times.size)

#Plot fourier transform of spherical harmonic coefficients during the orbit.
#General behavior seems to be that ell=1 distortions vary at 1 year^-1 frequency, ell = 2 at 1/2 year^-1, etc.
plt.plot(freqs[:20], np.abs(psi10_fft_time)[:20], label = r'$\Psi_{10}$')
plt.plot(freqs[:20], np.abs(psi11_fft_time)[:20], label = r'$\Psi_{11}$')
plt.plot(freqs[:20], np.abs(psi1n1_fft_time)[:20], label = r'$\Psi_{1-1}$')
plt.plot(freqs[:20], np.abs(psi20_fft_time)[:20], label = r'$\Psi_{20}$')
plt.plot(freqs[:20], np.abs(psi21_fft_time)[:20], label = r'$\Psi_{21}$')
plt.plot(freqs[:20], np.abs(psi2n1_fft_time)[:20], label = r'$\Psi_{2-1}$')
plt.title(r'Abs. value of FFT of amplitudes')
plt.legend()
plt.yscale('log')
plt.ylabel('Amplitude (radians)')
plt.xlabel('Frequency (Hz)')
```
